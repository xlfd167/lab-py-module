```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])

print(a)
print(type(a))
print("shape:", a.shape)
print("dtype:", a.dtype)
```

```python
## 数组创建方法
print(np.zeros((2, 3)))
print(np.ones((2, 3)))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
```

```python
## 向量化计算
a = np.array([1, 2, 3, 4, 5])

print("a + 10 =", a + 10)
print("a * 2  =", a * 2)
print("a ** 2 =", a ** 2)
```

```python
## 索引与切片
a = np.arange(10)

print("完整数组:", a)
print("第一个元素:", a[0])
print("前三个:", a[:3])
print("偶数位置:", a[::2])
print("大于 5:", a[a > 5])
```

```python
## 二维数组与统计
m = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("shape:", m.shape)
print("总和:", m.sum())
print("平均值:", m.mean())
print("按列求和:", m.sum(axis=0))
print("按行求和:", m.sum(axis=1))
```

```python
## 矩阵乘法
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(A @ B)
```
