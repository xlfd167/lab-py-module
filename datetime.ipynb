```python
from datetime import date, time, datetime, timedelta

today = date.today()
print(f"today: {today}")

now = datetime.now()
print(f"now: {now}")
```

```python
## 格式化输出
print(now.strftime("%Y-%m-%d %H:%M:%S"))
```

```python
## string 转 datetime
text = "2026-09-29 20:30:00"
dt = datetime.strptime(text, "%Y-%m-%d %H:%M:%S")
print(dt)
```
