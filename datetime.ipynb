```python
from datetime import date, time, datetime, timedelta

today = date.today()
print(f"today: {today}")

now = datetime.now()
print(f"now: {now}")

dt = datetime(year=2026,
              month=10,
              day=24,
              hour=12,
              minute=34,
              second=56,
              )
print(dt)
```

```python
## 格式化输出
print(now.strftime("%Y-%m-%d %H:%M:%S"))
```

```python
## string 转 datetime
text = "2026-09-29 20:30:00"
dt = datetime.strptime(text, "%Y-%m-%d %H:%M:%S")
print(dt)
```

```python
# timedelta：时间加减
now = datetime.now()

tomorrow = now + timedelta(days=1)
one_hour_ago = now - timedelta(hours=7)

print(f"tomorrow: {tomorrow}")
print(f"one_hour_ago: {one_hour_ago}")

```
